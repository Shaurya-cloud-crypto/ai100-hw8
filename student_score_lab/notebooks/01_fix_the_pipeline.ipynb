## Step 10 - Scoreboard
Trains both models on the current pipeline and logs the result. **Re-run after every patch** (Restart & Run All).

To start the scoreboard over, set `RESET_SCOREBOARD = True` once.

````python
RESET_SCOREBOARD = False

lin = LinearRegression().fit(X_train_model, y_train_score)
pred_score = lin.predict(X_test_model)

with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter("always")
    logit = LogisticRegression().fit(X_train_model, y_train_pass)
p_pass = logit.predict_proba(X_test_model)[:, 1]
if any("converge" in str(w.message) for w in caught):
    print("⚠️  Logistic regression's optimizer stopped before it finished (ConvergenceWarning).")

fails = (y_test_pass == 0).values
label = "no patches" if not PATCHES else "patches " + ",".join(map(str, sorted(PATCHES)))
row = {"run": label, "students": len(df), "features": len(FEATURES),
       "R2": r2_score(y_test_score, pred_score), "MAE": mean_absolute_error(y_test_score, pred_score),
       "accuracy": accuracy_score(y_test_pass, p_pass >= 0.5),
       "fails_found": f"{((p_pass < 0.5) & fails).sum()} of {fails.sum()}",
       "AUC": roc_auc_score(y_test_pass, p_pass)}

log_file = OUT / "scoreboard.csv"
board = pd.read_csv(log_file) if log_file.exists() and not RESET_SCOREBOARD else pd.DataFrame()
if len(board):
    board = board[board["run"] != label]
board = pd.concat([board, pd.DataFrame([row])], ignore_index=True)
board.to_csv(log_file, index=False)

print(f"This run: {label}   ({len(PATCHES)} of 9 patches)\n")
board.round(3)
````

````python
fig, axes = plt.subplots(1, 2, figsize=(11, 0.45 * len(board) + 1.6), sharey=True)
for ax, metric, goal in [(axes[0], "R2", 0.87), (axes[1], "AUC", 0.97)]:
    colors = ["#c0392b" if r == label else "#2a6fb0" for r in board["run"]]
    ax.barh(board["run"], board[metric], color=colors)
    ax.axvline(goal, ls="--", color="gray"); ax.text(goal, -0.6, f" goal ≈ {goal}", color="gray", va="top")
    lo = max(0, board[metric].min() - 0.05)
    ax.set_xlim(lo, 1.0); ax.set_title(f"{metric} (test set)")
    for y, v in enumerate(board[metric]):
        ax.text(v, y, f" {v:.3f}", va="center", fontsize=8)
    ax.spines[["top", "right"]].set_visible(False)
axes[0].invert_yaxis()
plt.suptitle("Scoreboard: every run you made (red = this run)", y=1.0)
plt.tight_layout(); plt.show()
````

**1. Which patch improved the scoreboard the most? Which did I predict wrong?**

For R², Patch 2 gave the biggest jump: 0.728 → 0.762 (+0.035), and MAE fell from 6.71 to 6.07 points. Before it, `on_time_rate` only counted the exact spelling `"on time"`, so most on-time homework was counted as not on time and the feature was mostly noise. Patch 6 (+0.032) and Patch 8 (+0.032) were close behind. For AUC the biggest jump was Patch 5 (0.910 → 0.935).

I predicted two wrong:

- **Patch 1.** I guessed "same" because 8 rows out of 408 looked too few to matter. R² actually went down, 0.752 → 0.728. The duplicates were making the test set easier: some students had one copy in train and one in test, so the model was being tested on rows it had already seen. The lower number is the honest one.
- **Patch 7.** I guessed "same" because the patch only removes columns, it adds no new information. R² went up a little, 0.832 → 0.841, and MAE fell from 4.97 to 4.83. Going from 14 features to 7 removed `seat_row` (pure noise) and the six single homework columns that repeat what `hw_avg` already says, so the model had less noise to fit by accident.

**2. Patch 7 off, `letter_grade` added as a number (F=0 … A=4)**

I ran this in a separate copy of the lab so it would not change my scoreboard. With patches 1-6, 8, 9 and `letter_grade` as a 15th feature I got **R² = 0.925** (MAE 3.17), and the classifier became perfect: accuracy 1.0, AUC 1.0, 26 of 26 fails found. That is much better than my real result (R² 0.873, AUC 0.972), but it is fake. `letter_grade` is calculated from `final_score`, so the model is being given the answer. In real use the letter grade does not exist until the final score is known, so this model could never be used to predict anything. This is target leakage. R² is not exactly 1 only because a letter is a coarse version of the score (5 bins for a 0-100 number).

**3. 120% attendance: blank or 100%?**

- *For blank (what Patch 5 does):* 120% is impossible, so we know it is wrong but not what the true value was. It could be a typo for 20% or 100% or 12%. Calling it 100% turns a broken record into a perfect one. Blank is honest, and Patch 8 then fills it with the train median.
- *For 100%:* if the cause is something like extra sessions being counted, the student really did attend a lot, and 100% keeps that information. The median fill would pull a top attender down to an average one. In my data only 4 values per attendance column are affected, so either choice changes little, but I would only cap at 100% if I knew how the error happened.
